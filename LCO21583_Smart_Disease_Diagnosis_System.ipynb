Encode Disease Name to Numbers

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

train_df['prognosis'] = encoder.fit_transform(train_df['prognosis'])
test_df['prognosis'] = encoder.transform(test_df['prognosis'])

disease_names = encoder.classes_
```

Split Features & Labels

```python
X_train = train_df.drop('prognosis', axis=1)
y_train = train_df['prognosis']

X_test = test_df.drop('prognosis', axis=1)
y_test = test_df['prognosis']
```

Train the Machine Learning Model

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)
```

Model Evaluation

```python
train_df = train_df.drop(columns=["Unnamed: 133"], errors="ignore")
test_df = test_df.drop(columns=["Unnamed: 133"], errors="ignore")
```

```python
print(train_df.shape)
print(test_df.shape)
```

```python
X_train = train_df.drop('prognosis', axis=1)
y_train = train_df['prognosis']

X_test = test_df.drop('prognosis', axis=1)
y_test = test_df['prognosis']
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Model Accuracy:", accuracy_score(y_test, y_pred))
```

Final Disease Prediction

```python
def smart_diagnosis(symptoms):
    input_data = np.zeros(len(X_train.columns))

    for symptom in symptoms:
        if symptom in X_train.columns:
            input_data[X_train.columns.get_loc(symptom)] = 1

    pred = model.predict([input_data])
    disease = encoder.inverse_transform(pred)[0]

    guidance = {
        "Diabetes": "Maintain low sugar diet, exercise daily, consult doctor.",
        "Hypertension": "Reduce salt, manage stress, monitor BP regularly.",
        "Common Cold": "Take rest, warm fluids, hydration."
    }

    return disease, guidance.get(disease, "Consult a medical professional.")
```

```python
symptoms = ["itching", "skin_rash", "fatigue"]
disease, advice = smart_diagnosis(symptoms)

print("Predicted Disease:", disease)
print("Health Guidance:", advice)
```
